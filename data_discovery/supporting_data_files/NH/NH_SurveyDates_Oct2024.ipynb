| Variable Name (Column Header) | Description | Variable Type |
|---|---|---|
| CMS Certification Number (CCN) | CMS Certification Number (CCN) | Text (6) |
| Survey Date | Date the survey was completed | Date |
| Survey Type | Type of survey conducted (Health, Fire Safety, Complaint, Infection Control) | Text |
| Survey Scope | Indicates whether the survey was a standard survey or a complaint survey | Text |
| Processing Date | Date the data were retrieved | Date |
